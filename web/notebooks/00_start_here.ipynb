# Data Make Sins

Seven ways numbers fool people. One notebook for each mistake.

Every notebook has the same shape:

1. The mistake in one sentence
2. A real case from the news
3. A small dataset, analysed with pandas
4. A figure, and an animated figure
5. A better method, when there is one
6. One question to ask next time

| # | Mistake | Real case | Notebook |
|---|---------|-----------|----------|
| 1 | Survivorship bias | The bombers that came back, 1943 | [01_survivorship_bias](01_survivorship_bias.ipynb) |
| 2 | Mean instead of median | US wealth: mean 551,000, median 108,000 | [02_mean_vs_median](02_mean_vs_median.ipynb) |
| 3 | Relative risk without absolute risk | Bacon and the 18 percent, WHO 2015 | [03_relative_vs_absolute_risk](03_relative_vs_absolute_risk.ipynb) |
| 4 | A chart axis that does not start at zero | Fox tax chart 2012, Georgia chart 2020 | [04_truncated_axis](04_truncated_axis.ipynb) |
| 5 | Correlation is not causation | Chocolate and Nobel prizes, 2012 | [05_correlation_vs_causation](05_correlation_vs_causation.ipynb) |
| 6 | Base rate neglect | The 99 percent accurate test, Cardiff 2017 | [06_base_rate_neglect](06_base_rate_neglect.ipynb) |
| 7 | Publication bias | 74 antidepressant trials, 2008 | [07_publication_bias](07_publication_bias.ipynb) |

## How to run

On your computer:

```
pip install -r requirements.txt
jupyter lab
```

Then open any notebook in this folder and run the cells from top to bottom.

On Google Colab: open a notebook, run the first cell, and the repo downloads itself.
